# Regression

1. What is Simple Linear Regression
   - Simple Linear Regression is a supervised Machine Learning algorithm used to predict a continuous dependent variable (Y) using one independent variable (X).

   The equation is:

   Y=mX+c

2. What are the key assumptions of Simple Linear Regression
   - The main assumptions are:

     - Linearity – There should be a linear relationship between X and Y.
     - Independence – Observations should be independent of each other.
     - Homoscedasticity – The variance of errors should remain approximately constant.
     - Normality of residuals – The residuals/errors should be approximately - - normally distributed, especially for inference.
     - No significant outliers – Extreme outliers should not unduly influence the regression line.

3. What does the coefficient m represent in the equation Y=mX+c
   - m represents the slope of the regression line.

     It tells us how much Y changes when X increases by one unit.

4. What does the intercept c represent in the equation Y=mX+c
   - c represents the intercept, which is the predicted value of Y when X is 0.

5. How do we calculate the slope m in Simple Linear Regression
   - The slope can be calculated using:

$$ m = \frac{\sum (X_i-\bar X)(Y_i-\bar Y)} {\sum (X_i-\bar X)^2} $$
       Where:

      (X_i\) = individual X value
      (Y_i\) = individual Y value
      (\bar X\) = mean of X
      (\bar Y\) = mean of Y

6.  What is the purpose of the least squares method in Simple Linear Regression
    - The least squares method is used to find the best-fitting regression line.It chooses the values of m and c that minimize the sum of squared differences between actual and predicted values.

7. How is the coefficient of determination (R²) interpreted in Simple Linear Regression
   - R² (coefficient of determination) tells us the proportion of variation in the dependent variable that is explained by the regression model.

     Its value is usually between 0 and 1.

8. What is Multiple Linear Regression
   - Multiple Linear Regression is used to predict a continuous dependent variable using two or more independent variables.

9. What is the main difference between Simple and Multiple Linear Regression
   - Simple Linear Regression
     - Uses one independent variable
     - Y = mX + c
     - Example: Salary based on experience
     - Easier to visualize

    - Multiple Linear Regression
      - Uses two or more independent variables
      - Y = b₀ + b₁X₁ + b₂X₂ + ...
      - Example: Salary based on experience, education, age, etc.
      - More complex

10. What are the key assumptions of Multiple Linear Regression
    - The main assumptions are:

      1. Linearity

         - There should be a linear relationship between the predictors and the dependent variable.

      2. Independence

         - Observations/errors should be independent.

      3. Homoscedasticity

         - The variance of the residuals should be approximately constant.

      4. Normality of Residuals

         - Residuals should be approximately normally distributed when normality is needed for statistical inference.

      5. No Severe Multicollinearity

         - The independent variables should not be highly correlated with each other.
      
11. What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model
   - Heteroscedasticity occurs when the variance of the residuals is not constant across different levels of the predicted values or predictors.

   - Effects:
     - Coefficient estimates can still be unbiased under standard conditions, but they are not efficient.
     - Standard errors can be incorrect.
     - Confidence intervals and hypothesis tests can become unreliable.
     - It can make us incorrectly conclude that a variable is statistically significant.

12. How can you improve a Multiple Linear Regression model with high multicollinearity
    - Multicollinearity occurs when independent variables are highly correlated with each other.

    - Common solutions:
      - Remove one of the highly correlated variables
      - Combine variables
      - Use regularization
      - Use PCA
      - Check VIF
13.  What are some common techniques for transforming categorical variables for use in regression models
    - Categorical variables need to be converted into numerical representations.
      - One-Hot Encoding
      - Ordinal Encoding
      - Dummy Variable Encoding
      - Target/Mean Encoding

14. What is the role of interaction terms in Multiple Linear Regression
    - An interaction term is used when the effect of one variable on the target depends on another variable.

15. How can the interpretation of intercept differ between Simple and Multiple Linear Regression
    - Simple Linear Regression
      $$ Y = mX + c $$

      The intercept c represents the predicted value of Y when X = 0.

    - Multiple Linear Regression
      $$ Y = b_0+b_1X_1+b_2X_2+...+b_nX_n $$

      The intercept represents the predicted value of Y when all predictors are equal to 0.

16. What is the significance of the slope in regression analysis, and how does it affect predictions
    - The slope/coefficient tells us how much the predicted value of Y changes when a predictor increases by one unit, holding other predictors constant in multiple regression.

     Example:

     $$ Salary = 25,000 + 4,000(Experience) $$

     The slope is 4,000.

     - Effect on predictions:
       - Positive slope → predicted Y increases
       - Negative slope → predicted Y decreases
       - Larger absolute slope → larger change in predicted Y per unit, but only relative to the variable's units/scaling

17.  How does the intercept in a regression model provide context for the relationship between variables
     - The intercept provides a baseline prediction.

      It tells us the expected value of the dependent variable when the predictor(s) are zero.

      For example:

      $$ Sales = 10 + 2(Advertising) $$

      The intercept 10 means that when advertising expenditure is zero, the model predicts sales of 10 units.

18. What are the limitations of using R² as a sole measure of model performance
    - R² is useful, but we should never use it as the only measure of model performance.

    - Limitations:
      - High R² doesn't necessarily mean a good model
      - R² generally increases when predictors are added
      - R² doesn't prove causation
      - It doesn't directly measure prediction error
      - A low R² isn't always bad

19. How would you interpret a large standard error for a regression coefficient
    - A standard error measures the uncertainty or variability of an estimated coefficient.

      A large standard error means:

      The estimated coefficient is relatively uncertain or imprecise.

      For example:
                 Coefficient = 10
                 Standard Error = 8

                 The coefficient estimate is relatively uncertain.

20. How can heteroscedasticity be identified in residual plots, and why is it important to address it
    - Heteroscedasticity can be identified by plotting the residuals against the predicted (fitted) values of a regression model.

      If the residuals are randomly scattered with approximately equal spread across all predicted values, the data is considered to have constant variance (homoscedasticity).

      However, if the spread of residuals increases or decreases systematically, such as forming a funnel or cone-shaped pattern, it indicates heteroscedasticity.

    - Why is it important to address?

      Heteroscedasticity is important because it can make the standard errors of regression coefficients unreliable. As a result, statistical tests, p-values, and confidence intervals may be misleading. It can also reduce the efficiency of the regression estimates.

      Therefore, detecting and addressing heteroscedasticity helps ensure that the statistical conclusions and inferences from the regression model are more reliable.

21. What does it mean if a Multiple Linear Regression model has a high R² but low adjusted R²
    - A high R² but low adjusted R² usually indicates that some of the predictors may not contribute meaningfully to the model. R² always increases or stays the same when new variables are added, even if they are irrelevant. Adjusted R² penalizes the model for adding unnecessary variables.

22. Why is it important to scale variables in Multiple Linear Regression
    - Scaling variables means converting features to a similar range or scale.

      It is important when the independent variables have very different units or ranges, because scaling can make model coefficients easier to compare and can improve numerical stability.

23. What is polynomial regression
    - Polynomial Regression is a regression technique used to model a non-linear relationship between the independent variable and dependent variable by including powers of the independent variable.

      For example:

       $$ Y = b_0 + b_1X + b_2X^2 + b_3X^3 + \epsilon $$

24. How does polynomial regression differ from linear regression
    - Linear Regression
      - Models a straight-line relationship
      - Uses X
      - Example: Y = b₀ + b₁X
      - Suitable for approximately linear relationships

    - Polynomial Regression
      - Models a curved/non-linear relationship
      - Uses X, X², X³, etc.
      - Example: Y = b₀ + b₁X + b₂X²
      - Suitable for curved relationships

25. When is polynomial regression used
    - Polynomial Regression is used when:

      -The relationship between X and Y is non-linear.
      - A scatter plot shows a curved pattern.
      - Linear regression does not adequately fit the data.
      - We want to capture patterns such as U-shaped, inverted U-shaped, or other smooth curves.

26. What is the general equation for polynomial regression
    - Y=b0​+b1​X+b2​X2+b3​X3+⋯+bn​Xn+ϵ

      Where:

      - Y = dependent variable
      - X = independent variable
      - b₀ = intercept
      - b₁, b₂, ..., bₙ = coefficients
      - n = degree of the polynomial
      - ε = error term

27. Can polynomial regression be applied to multiple variables
    - Yes. Polynomial features can be created from multiple independent variables.

28. What are the limitations of polynomial regression
    - The main limitations are:
      - Overfitting – Higher-degree polynomials can fit the training data too closely.
      - Complexity – The model becomes difficult to interpret as the degree increases.
      - Poor extrapolation – Predictions outside the training range can become unrealistic.
      - Multicollinearity – Polynomial features such as X, X², and X³ can be highly correlated.
      - Sensitive to outliers – Outliers can strongly influence the fitted curve.

29. What methods can be used to evaluate model fit when selecting the degree of a polynomial
    - R² – Measures the proportion of variation explained by the model.
    - Adjusted R² – Penalizes unnecessary predictors and is useful for comparing models with different numbers of terms.
    - MSE (Mean Squared Error) – Measures average squared prediction error.
    - RMSE (Root Mean Squared Error) – Measures prediction error in the same units as the target.
    - MAE (Mean Absolute Error) – Measures average absolute prediction error.
    - Cross-validation – Helps determine whether the model generalizes well to unseen data.
    - Residual plots – Help identify systematic patterns or poor model fit.

30. Why is visualization important in polynomial regression.
    - Visualization helps us understand the relationship between the variables and see whether a polynomial curve fits the data well.

31. How is polynomial regression implemented in Python?
    - Polynomial Regression can be implemented using Scikit-learn's PolynomialFeatures and LinearRegression.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# Sample data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([2, 5, 10, 17, 26])

# Create polynomial features
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

# Create and train model
model = LinearRegression()
model.fit(X_poly, y)

# Predictions
y_pred = model.predict(X_poly)

# Plot
plt.scatter(X, y)
plt.plot(X, y_pred)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Polynomial Regression")
plt.show()

print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
```
